# Studi Kasus: Dashboard Analisis Penjualan


## 1. Instalasi dan Dataset

```python
!pip -q install pandas numpy plotly streamlit
import pandas as pd
import numpy as np
import plotly.express as px
from IPython.display import display

np.random.seed(42)
tanggal = pd.date_range("2025-01-01", periods=180, freq="D")
kategori = ["Elektronik", "Pakaian", "Makanan", "Peralatan Rumah"]

df = pd.DataFrame({
    "Tanggal": np.random.choice(tanggal, 600),
    "Kategori": np.random.choice(kategori, 600),
    "Jumlah": np.random.randint(1, 8, 600),
    "Harga": np.random.choice([25000, 50000, 75000, 100000, 150000, 250000], 600)
})
df["Pendapatan"] = df["Jumlah"] * df["Harga"]
df = df.sort_values("Tanggal").reset_index(drop=True)
df.head()
```

## 2. Ringkasan Data

```python
display(df.describe(include="all"))
print("Total pendapatan:", f"Rp{df['Pendapatan'].sum():,.0f}")
print("Jumlah transaksi:", len(df))
```

## 3. Visualisasi

```python
harian = df.groupby("Tanggal", as_index=False)["Pendapatan"].sum()
fig = px.line(harian, x="Tanggal", y="Pendapatan", title="Tren Pendapatan Harian")
fig.update_layout(xaxis_title="Tanggal", yaxis_title="Pendapatan (Rp)")
fig.show()

kategori_df = df.groupby("Kategori", as_index=False)["Pendapatan"].sum().sort_values("Pendapatan", ascending=False)
fig = px.bar(kategori_df, x="Kategori", y="Pendapatan", title="Pendapatan per Kategori",
             text_auto=".2s")
fig.update_layout(xaxis_title="Kategori", yaxis_title="Pendapatan (Rp)")
fig.show()

fig = px.scatter(df, x="Jumlah", y="Pendapatan", color="Kategori",
                 title="Hubungan Jumlah Barang dan Pendapatan")
fig.show()
```

## 4. Filter Kategori

```python
pilihan = "Elektronik"  # Ganti kategori sesuai kebutuhan
hasil_filter = df[df["Kategori"] == pilihan]
display(hasil_filter.head(10))
print("Pendapatan kategori:", f"Rp{hasil_filter['Pendapatan'].sum():,.0f}")
```

## 5. Membuat Aplikasi Streamlit

```python
%%writefile dashboard_app.py
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

st.set_page_config(page_title="Dashboard Penjualan", layout="wide")
st.title("Dashboard Analisis Tren Penjualan")
st.caption("Data simulasi | Filter interaktif dan visualisasi sederhana")

@st.cache_data
def buat_data():
    np.random.seed(42)
    tanggal = pd.date_range("2025-01-01", periods=180, freq="D")
    kategori = ["Elektronik", "Pakaian", "Makanan", "Peralatan Rumah"]
    df = pd.DataFrame({
        "Tanggal": np.random.choice(tanggal, 600),
        "Kategori": np.random.choice(kategori, 600),
        "Jumlah": np.random.randint(1, 8, 600),
        "Harga": np.random.choice([25000, 50000, 75000, 100000, 150000, 250000], 600)
    })
    df["Pendapatan"] = df["Jumlah"] * df["Harga"]
    return df.sort_values("Tanggal")

df = buat_data()
c1, c2 = st.columns(2)
with c1:
    kategori_pilih = st.multiselect("Kategori", sorted(df["Kategori"].unique()),
                                    default=sorted(df["Kategori"].unique()))
with c2:
    rentang = st.date_input("Rentang tanggal",
                            value=(df["Tanggal"].min().date(), df["Tanggal"].max().date()))

if len(rentang) == 2:
    mulai, akhir = rentang
    data = df[
        df["Kategori"].isin(kategori_pilih) &
        (df["Tanggal"].dt.date >= mulai) &
        (df["Tanggal"].dt.date <= akhir)
    ]
else:
    data = df[df["Kategori"].isin(kategori_pilih)]

m1, m2, m3 = st.columns(3)
m1.metric("Total Pendapatan", f"Rp{data['Pendapatan'].sum():,.0f}")
m2.metric("Transaksi", f"{len(data):,}")
m3.metric("Total Barang", f"{data['Jumlah'].sum():,}")

if data.empty:
    st.warning("Tidak ada data untuk filter yang dipilih.")
else:
    harian = data.groupby("Tanggal", as_index=False)["Pendapatan"].sum()
    st.plotly_chart(px.line(harian, x="Tanggal", y="Pendapatan",
                            title="Tren Pendapatan Harian"), use_container_width=True)

    per_kategori = data.groupby("Kategori", as_index=False)["Pendapatan"].sum()
    st.plotly_chart(px.bar(per_kategori, x="Kategori", y="Pendapatan",
                           title="Pendapatan per Kategori", text_auto=".2s"),
                    use_container_width=True)

    st.plotly_chart(px.scatter(data, x="Jumlah", y="Pendapatan", color="Kategori",
                               title="Jumlah Barang vs Pendapatan"),
                    use_container_width=True)

    st.subheader("Data Transaksi")
    st.dataframe(data, use_container_width=True)
    st.download_button("Unduh CSV", data.to_csv(index=False).encode("utf-8"),
                       file_name="data_penjualan.csv", mime="text/csv")
```
